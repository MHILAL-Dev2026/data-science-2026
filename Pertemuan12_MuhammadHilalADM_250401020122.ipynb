**Langkah 1: Generate & Eksplorasi Dataset Transaksi**

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
produk = ['Roti', 'Selai', 'Susu', 'Sereal', 'Telur',
          'Keju', 'Kopi', 'Gula', 'Teh', 'Mentega']

# Buat 50 transaksi, tiap transaksi berisi 2-5 produk
transaksi = []
for _ in range(50):
    n_item = np.random.randint(2, 6)
    transaksi.append(list(np.random.choice(produk, n_item, replace=False)))

# Suntikkan pola: Roti sering bersama Selai
for i in range(0, 20):
    if 'Roti' in transaksi[i] and 'Selai' not in transaksi[i]:
        transaksi[i].append('Selai')

print('Contoh transaksi:', transaksi[:3])
print('Jumlah transaksi:', len(transaksi))
```

**Langkah 2: One-Hot Encoding Transaksi**

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(transaksi).transform(transaksi)
df = pd.DataFrame(te_ary, columns=te.columns_)
print(df.head())
```

**Langkah 3: Cari Frequent Itemset dengan Apriori**

```python
from mlxtend.frequent_patterns import apriori

for ms in [0.05, 0.1, 0.2]:
    freq = apriori(df, min_support=ms, use_colnames=True)
    print(f'min_support={ms}: {len(freq)} itemset ditemukan')

# Gunakan min_support yang menghasilkan jumlah itemset wajar (tidak 0, tidak ratusan)
freq_items = apriori(df, min_support=0.1, use_colnames=True)
freq_items = freq_items.sort_values('support', ascending=False)
print(freq_items.head(10))
```

**Langkah 4: Bentuk & Saring Aturan Asosiasi**

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(freq_items, metric='confidence',
                          min_threshold=0.5)
rules = rules[rules['lift'] > 1].sort_values('lift', ascending=False)

print(rules[['antecedents', 'consequents',
             'support', 'confidence', 'lift']].head(10))
```

Interpretasi:

1. Aturan mana yang paling kuat (Lift tertinggi)?
    
    Sesuai dengan kode yang menyuntikkan pola 'Roti' sering bersama 'Selai' pada Langkah 1, aturan asosiasi dengan Lift tertinggi dipastikan melibatkan kumpulan item (Roti) -> (Selai)

2. Apakah masuk akal secara bisnis (mis. Roti -> Selai)?

    masuk akal secara logika bisnis. Roti dan Selai adalah barang komplementer (saling melengkapi) yang memang secara praktis sering dibeli bersamaan untuk konsumsi harian

**Langkah 5: Rekomender Sederhana dengan Content-Based Filtering**

```python
from sklearn.metrics.pairwise import cosine_similarity

katalog = pd.DataFrame({
    'produk': produk,
    'kategori': ['Bakery', 'Bakery', 'Dairy', 'Bakery', 'Dairy',
                 'Dairy', 'Minuman', 'Bumbu', 'Minuman', 'Dairy']
})

fitur = pd.get_dummies(katalog['kategori'])
sim_matrix = cosine_similarity(fitur)

def rekomendasi_serupa(nama_produk, top_n=3):
    idx = katalog.index[katalog['produk'] == nama_produk][0]
    skor = list(enumerate(sim_matrix[idx]))
    skor = sorted(skor, key=lambda x: x[1], reverse=True)
    skor = [s for s in skor if s[0] != idx][:top_n]
    return katalog.iloc[[i for i, _ in skor]]['produk'].tolist()

print('Mirip dengan Roti:', rekomendasi_serupa('Roti'))
```

**Langkah 6: Bandingkan Kedua Pendekatan**

```python
produk_target = 'Roti'

# Dari association rules: cari consequents dari aturan yang antecedent-nya mengandung produk_target
rules_terkait = rules[rules['antecedents'].apply(
    lambda x: produk_target in x)]
print('Rekomendasi dari Association Rules:')
print(rules_terkait[['consequents', 'lift']].head())

print('Rekomendasi dari Content-Based:', rekomendasi_serupa(produk_target))
```

Interpretasi:

1. Diskusikan: apakah kedua pendekatan memberi rekomendasi yang konsisten?

    Rekomendasi ini wajar untuk tidak konsisten karena fokus algoritmanya berbeda. Association rules mencari pola keranjang belanja untuk barang komplementer, seperti penemuan kombinasi Roti dan Selai. Pendekatan Content-Based Filtering mencari barang yang ekuivalen berdasarkan atribut kategori substitusi, seperti kombinasi Roti dan Sereal yang masuk dalam kategori yang sama

2. Kapan sebaiknya menggunakan salah satu, atau menggabungkan keduanya (hybrid)?

    -Association Rules: Algoritma ini ideal diterapkan pada halaman keranjang belanja untuk mendukung strategi penawaran silang (cross-selling)
    
      -Content-Based Filtering: Metode ini direkomendasikan untuk menghindari masalah cold start saat sistem baru dan belum memiliki akumulasi data riwayat transaksi pelanggan
      
      -Hybrid: Strategi gabungan sangat direkomendasikan agar bisa memberikan penawaran barang pengganti dan barang pelengkap secara simultan yang menguntungkan pelanggan

# Kesimpulan

Pendekatan Association Rules (Apriori) dan Content-Based Filtering memberikan hasil yang berbeda namun pada dasarnya saling melengkapi. Association Rules berfokus pada pola perilaku belanja nyata untuk menemukan produk komplementer yang sering dibeli secara bersamaan (seperti hubungan kuat antara Roti dan Selai), sehingga metode ini sangat efektif digunakan untuk strategi cross-selling di halaman keranjang belanja. Sebaliknya, Content-Based Filtering merekomendasikan produk substitusi berdasarkan kemiripan atribut atau kategorinya (seperti Roti dan Sereal yang sama-sama Bakery), yang sangat berguna saat pengguna mengeksplorasi alternatif produk atau untuk mengatasi masalah cold start ketika data riwayat transaksi belum memadai. Oleh karena itu, penggabungan kedua metode ini menjadi sebuah sistem hybrid merupakan strategi paling ideal, karena memungkinkan bisnis menyajikan rekomendasi barang pelengkap sekaligus barang pengganti secara komprehensif guna memaksimalkan pengalaman berbelanja dan meningkatkan potensi penjualan.
